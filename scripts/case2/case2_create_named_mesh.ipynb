## Named mesh
Define named physical groups in the mesh.

```python
import sys
import gmsh

gmsh.initialize()

# Load your existing mesh
gmsh.open('../../meshes/coil_box.msh')

# Define your mapping: {Material_ID: "Name"}
# dim=3 for volumes, dim=2 for surfaces
material_mapping = {
    1: "vacuum",
    2: "wire"
}

# Apply the names
for mat_id, name in material_mapping.items():
    gmsh.model.addPhysicalGroup(dim=3, tags=[mat_id], tag=mat_id, name=name)

gmsh.model.addPhysicalGroup(dim=2, tags=[8], tag=8, name="CoilIn")
gmsh.model.addPhysicalGroup(dim=2, tags=[11], tag=11, name="CoilOut")

gmsh.model.addPhysicalGroup(dim=2, tags=[1], tag=1, name="Side1")
gmsh.model.addPhysicalGroup(dim=2, tags=[2], tag=2, name="Side2")
gmsh.model.addPhysicalGroup(dim=2, tags=[3], tag=3, name="Side3")
gmsh.model.addPhysicalGroup(dim=2, tags=[4], tag=4, name="Side4")
gmsh.model.addPhysicalGroup(dim=2, tags=[5], tag=5, name="Side5")
gmsh.model.addPhysicalGroup(dim=2, tags=[12], tag=12, name="Side6")

# Save the mesh
# Use Gmsh 2.2 not 4, otherwise MOOSE would complain
gmsh.option.setNumber("Mesh.MshFileVersion", 2.2)
gmsh.write('../../meshes/coil_box_named.msh')

gmsh.finalize()
```
