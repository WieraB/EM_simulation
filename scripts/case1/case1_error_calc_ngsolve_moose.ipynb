```python
import numpy as np
import pyvista as pv
from scipy.spatial import KDTree

case_name = "case1"
mesh_name = "coil_box"

sol_ngsolve = pv.read(f"../../output/{case_name}/{case_name}_ngsolve.vtu")
elec_pot_ngsolve = sol_ngsolve["electric_potential"]
curr_dens_ngsolve = sol_ngsolve["current_density"]

sol_moose = pv.read(f"../../output/{case_name}/{case_name}_moose.vtu")
elec_pot_moose = sol_moose["electric_potential"]
curr_dens_moose = sol_moose["current_density"]

print(elec_pot_ngsolve.shape)
print(elec_pot_moose.shape)

print(curr_dens_ngsolve.shape)
print(curr_dens_moose.shape)
```

```python
mesh = pv.read(f"../../meshes/coil_box_named.msh")
```

```python
# NGSolve and MOOSE

abs_error = np.abs(elec_pot_ngsolve - elec_pot_moose)

abs_error_max = np.max(abs_error)
abs_error_mean = np.mean(abs_error)

print (f"Case: {case_name}")
print("")
print(f"  * Max. absolute error between NGSolve and MOOSE : {abs_error_max:.3e}.")
print(f"  * Avg. absolute error between NGSolve and MOOSE : {abs_error_mean:.3e}.")
print("")

mesh["abs_error_elec_pot_ngsolve_moose"] = abs_error



rel_error = np.abs(elec_pot_ngsolve - elec_pot_moose) * 100 / np.abs(elec_pot_moose)

rel_error_max = np.max(rel_error[np.isfinite(rel_error)])
rel_error_mean = np.mean(rel_error[np.isfinite(rel_error)])

print (f"Case: {case_name}")
print("")
print(f"  * Max. relative error between NGSolve and MOOSE : {rel_error_max:.3e} %.")
print(f"  * Avg. relative error between NGSolve and MOOSE : {rel_error_mean:.3e} %.")
print("")

mesh["rel_error_elec_pot_ngsolve_moose"] = rel_error
```

```python
# NGSolve and MOOSE

curr_dens_ngsolve_x = curr_dens_ngsolve[:, 0]
curr_dens_moose_x = curr_dens_moose[:, 0]

abs_error_x = np.abs(curr_dens_ngsolve_x - curr_dens_moose_x)

abs_error_max = np.max(abs_error_x)
abs_error_mean = np.mean(abs_error_x)

print (f"Case: {case_name}")
print("")
print(f"  * Max. absolute error between NGSolve and MOOSE : {abs_error_max:.3e}.")
print(f"  * Avg. absolute error between NGSolve and MOOSE : {abs_error_mean:.3e}.")
print("")

eps = 1e-6
denom = np.abs(curr_dens_moose_x)
num = np.abs(curr_dens_ngsolve_x - curr_dens_moose_x)
rel_error_x = np.where(denom > eps, num * 100 / denom, np.nan)

rel_error_max = np.max(rel_error_x[np.isfinite(rel_error_x)])
rel_error_mean = np.mean(rel_error_x[np.isfinite(rel_error_x)])

print (f"Case: {case_name}")
print("")
print(f"  * Max. relative error between NGSolve and MOOSE : {rel_error_max:.3e} %.")
print(f"  * Avg. relative error between NGSolve and MOOSE : {rel_error_mean:.3e} %.")
print("")
```

```python
# NGSolve and MOOSE

curr_dens_ngsolve_y = curr_dens_ngsolve[:, 1]
curr_dens_moose_y = curr_dens_moose[:, 1]

abs_error_y = np.abs(curr_dens_ngsolve_y - curr_dens_moose_y)

abs_error_max = np.max(abs_error_y)
abs_error_mean = np.mean(abs_error_y)

print (f"Case: {case_name}")
print("")
print(f"  * Max. absolute error between NGSolve and MOOSE : {abs_error_max:.3e}.")
print(f"  * Avg. absolute error between NGSolve and MOOSE : {abs_error_mean:.3e}.")
print("")

eps = 1e-6
denom = np.abs(curr_dens_moose_y)
num = np.abs(curr_dens_ngsolve_y - curr_dens_moose_y)
rel_error_y = np.where(denom > eps, num * 100 / denom, np.nan)

rel_error_max = np.max(rel_error_y[np.isfinite(rel_error_y)])
rel_error_mean = np.mean(rel_error_y[np.isfinite(rel_error_y)])

print (f"Case: {case_name}")
print("")
print(f"  * Max. relative error between NGSolve and MOOSE : {rel_error_max:.3e}.")
print(f"  * Avg. relative error between NGSolve and MOOSE : {rel_error_mean:.3e}.")
print("")
```

```python
# NGSolve and MOOSE

curr_dens_ngsolve_z = curr_dens_ngsolve[:, 2]
curr_dens_moose_z = curr_dens_moose[:, 2]

abs_error_z = np.abs(curr_dens_ngsolve_z - curr_dens_moose_z)

abs_error_max = np.max(abs_error_z)
abs_error_mean = np.mean(abs_error_z)

print (f"Case: {case_name}")
print("")
print(f"  * Max. absolute error between NGSolve and MOOSE : {abs_error_max:.3e}.")
print(f"  * Avg. absolute error between NGSolve and MOOSE : {abs_error_mean:.3e}.")
print("")

eps = 1e-6
denom = np.abs(curr_dens_moose_z)
num = np.abs(curr_dens_ngsolve_z - curr_dens_moose_z)
rel_error_z = np.where(denom > eps, num * 100 / denom, np.nan)

rel_error_max = np.max(rel_error_z[np.isfinite(rel_error_z)])
rel_error_mean = np.mean(rel_error_z[np.isfinite(rel_error_z)])

print (f"Case: {case_name}")
print("")
print(f"  * Max. relative error between NGSolve and MOOSE : {rel_error_max:.3e}.")
print(f"  * Avg. relative error between NGSolve and MOOSE : {rel_error_mean:.3e}.")
print("")
```

```python
abs_error_curr_dens = np.zeros((len(abs_error), 3))
abs_error_curr_dens[:, 0] = abs_error_x
abs_error_curr_dens[:, 1] = abs_error_y
abs_error_curr_dens[:, 2] = abs_error_z

mesh["abs_error_curr_dens_ngsolve_moose"] = abs_error_curr_dens



rel_error_curr_dens = np.zeros((len(abs_error), 3))
rel_error_curr_dens[:, 0] = rel_error_x
rel_error_curr_dens[:, 1] = rel_error_y
rel_error_curr_dens[:, 2] = rel_error_z

mesh["rel_error_curr_dens_ngsolve_moose"] = rel_error_curr_dens
```
